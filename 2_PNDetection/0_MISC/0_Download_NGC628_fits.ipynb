```python
import pyvo

service = pyvo.dal.TAPService("https://archive.eso.org/tap_obs")

query = """
SELECT TOP 100
    target_name,
    dp_id,
    dataproduct_type,
    dataproduct_subtype,
    s_resolution,
    access_estsize
FROM ivoa.ObsCore
WHERE obs_collection = 'PHANGS'
  AND target_name LIKE '%NGC%628%'
ORDER BY dp_id
"""

result = service.search(query)
table = result.to_table()

table
```

```python
query = """
SELECT
    product_id,
    archive_id,
    original_filename,
    eso_category,
    access_estsize,
    access_url
FROM phase3v2.product_files
WHERE product_id = 'ADP.2021-07-16T10:20:56.333'
   OR product_id = 'ADP.2021-07-16T10:20:56.339'
ORDER BY product_id, archive_id
"""

result = service.search(query)
table = result.to_table()

table
```
