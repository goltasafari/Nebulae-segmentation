```python
from pathlib import Path
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
models = pd.read_csv("models_constrained_OI.csv")

pn_models = models[models["class"] == "PN"].reset_index(drop=True)
dig_models = models[models["class"] == "DIG"].reset_index(drop=True)
hii_models = models[models["class"] == "HII"].reset_index(drop=True)

print("# PN:   ", len(pn_models))
print("# HII:   ", len(hii_models))
print("# DIG:   ", len(dig_models))
```

```python
hbeta_halpha = 1/dig_models["H__1__4861A"]
nii_halpha = dig_models["N__2__6584A"]
oiii_halpha = dig_models["O__3__5007A"]

print("hbeta/halpha", np.min(hbeta_halpha), np.max(hbeta_halpha))
print("nii/halpha", np.min(nii_halpha), np.max(nii_halpha))
print("oiii/halpha", np.min(oiii_halpha), np.max(oiii_halpha))

plt.figure(figsize=(9, 5))
plt.hist(hbeta_halpha, bins=25,color='#3498db', edgecolor='#2c3e50', alpha=0.85, density=False)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
plt.figure(figsize=(9, 5))
plt.hist(nii_halpha, bins=25,color='#3498db', edgecolor='#2c3e50', alpha=0.85, density=False)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
plt.figure(figsize=(9, 5))
plt.hist(oiii_halpha, bins=25,color='#3498db', edgecolor='#2c3e50', alpha=0.85, density=False)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
