```python
from astropy.io import fits
import numpy as np

input_fits = "/Users/golta/Neb_det/data/NGC628_MAPS.fits"
output_fits = "NGC628_4channel.fits"

wanted_lines = [
    "HB4861",
    "OIII5006",
    "HA6562",
    "NII6583"]

hdul = fits.open(input_fits, memmap=True)

Hb   = hdul["HB4861_FLUX"].data
OIII = hdul["OIII5006_FLUX"].data
Ha   = hdul["HA6562_FLUX"].data
NII  = hdul["NII6583_FLUX"].data

data = np.stack([Hb, OIII, Ha, NII], axis=-1)
fits.writeto(output_fits, data, overwrite=True)
hdul.close()

print(data.shape)
```
